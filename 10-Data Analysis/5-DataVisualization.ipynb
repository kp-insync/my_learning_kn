```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

#create a line plot
plt.plot(x,y)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Basic Line Plot')
```

```python
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

# Creating customized line plot
plt.plot(x, y, color='red', linestyle='--', marker='o', linewidth=3, markersize=10)
plt.grid(True)
```

```python
# multiple plots
# sample data
x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
y2 = [1, 2, 3, 4, 5]

plt.figure(figsize=(10, 5))
plt.subplot(2, 2, 1) # one row, two columns, first plot
plt.plot(x, y1, color='green')
plt.title('Plot 1')

plt.subplot(2, 2, 2) # one row, two columns, second plot
plt.plot(y1, x, color='red')
plt.title('Plot 2')

plt.subplot(2, 2, 3) # one row, two columns, Third plot
plt.plot(x, y2, color='blue')
plt.title('Plot 3')

plt.subplot(2, 2, 4) # one row, two columns, fourth plot
plt.plot(x, y2, color='orange')
plt.title('Plot 4')
```
