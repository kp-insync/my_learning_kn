```python
# Remove a key from a dictionary using both pop() and del methods and explain the difference (in comments).
student = {'name': 'Kp', 'age': 31, 'grade': 'A'}

student['city'] = 'Bangalore'

student.pop('city')
print(student)

student['city'] = 'Bangalore'
del(student['city'])

"""
Pop - Used to remove the key from dictionary and it returns the deleted value
del - Used to remove the key from dictionary and it doesn't return the deleted value
"""
```

```python
# Write a program to find the fruit with the maximum price.
prices = {'apple': 100, 'banana': 40, 'cherry': 150}

max_price = max(list(prices.values()))

for fruit, price in prices.items():
    if price == max_price:
        print(fruit)
```

```python
# Write a program to iterate through a dictionary and print keys, values, and items using three different methods.
prices = {'apple': 100, 'banana': 40, 'cherry': 150}

for k in prices:
    print(k)

for v in prices.values():
    print(v)

for k, v in prices.items():
    print(k)
    print(v)
```

```python
# Write a program to create a new dictionary of subjects where marks > 80.
marks = {'Math': 88, 'Science': 92, 'English': 75, 'History': 89}

new_dict = {}

for sub, mark in marks.items():
    if marks[sub] > 80:
        new_dict[sub] = mark

print(new_dict)
```

```python
# Write a Python program to merge two dictionaries and handle duplicate keys by keeping the largest value.

dict_1 = {'a': 1, 'b': 2}
dict_2 = {'b': 3, 'c': 5}

merged_dict = {}

merged_dict = dict_1.copy()

for k, v in dict_2.items():
    if k not in merged_dict.keys():
        merged_dict[k] = v
    else:
        if k in merged_dict.keys():
            if merged_dict[k] > v:
                continue
            else:
                merged_dict[k] = v

print(merged_dict)
```

```python
# Write a program to swap keys and values in a dictionary.

votes = {'Alice': 5, 'Bob': 9}

swapped_votes = {v: k for k, v in votes.items()}

swapped_votes
```
